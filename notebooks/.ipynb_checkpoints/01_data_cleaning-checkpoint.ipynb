```python
payments=payments[payments['payment_type']!='not_defined']
```

```python
payments.info()
```

```python
payments_agg=payments.groupby('order_id').agg(
    total_payment_value=('payment_value','sum'),
    payment_installments=('payment_installments','sum'),
    payments_method_count=('payment_type','nunique'),
    primary_payment_type=('payment_type','first')
).reset_index()
```

```python
payments_agg.head()
```

## Exploring reviews dataset

```python
reviews.head()
```

```python
reviews.isna().sum()
```

```python
for col in ['review_creation_date', 'review_answer_timestamp']:
    reviews[col] = pd.to_datetime(reviews[col], errors='coerce')
```

## Exploring products dataset

```python
products.head(3)
```

```python
products['product_category_name'].nunique()
```

```python
eng_products.head(3)
```

```python
eng_products['product_category_name'].nunique()
```

```python
products=products.merge(eng_products,on='product_category_name',how='left')
```

```python
products.head(3)
```

```python
products.rename(columns={'product_category_name_english':'product_category'},inplace=True)
```

```python
products.isna().sum()
```

```python
products[products['product_category_name'].isnull()]
```

```python
products.dropna(subset={'product_category'},inplace=True)
```

## Creating a Master Dataset

```python
master=(orders
        .merge(items,on='order_id',how='inner')
        .merge(customers,on='customer_id',how='left')
        .merge(payments_agg,on='order_id',how='left')
        .merge(reviews[['order_id','review_score','review_comment_title','review_comment_message']],on='order_id',how='left')
        .merge(products[['product_id','product_category']],on='product_id',how='left'))
```

```python
master.head(3)
```

```python
master.isna().sum()
```

```python
master['total_payment_value'].fillna(0, inplace=True)
master['payment_installments'].fillna(0, inplace=True)
master['payments_method_count'].fillna(0, inplace=True)
master['primary_payment_type'].fillna('unknown', inplace=True)
```

```python
master.shape
```

```python
master[master['order_approved_at'].isnull()]
```

```python
master['order_approved_at'].fillna(master['order_purchase_timestamp'], inplace=True)
```

```python
master['is_delivered']=master['order_delivered_customer_date'].notna().astype(int)
```

```python
master['product_category'].fillna('unkown',inplace=True)
```

```python
master.isna().sum()
```

```python
master['review_score'].fillna(0,inplace=True)
```

```python
master.to_csv('../datasets/olist_master_dataset.csv',index=False)
```
