# Loading Dataset

```python
import pandas as pd
```

```python
df=pd.read_csv('../datasets/olist_master_dataset.csv')
df.columns
```

# Getting items per order

```python
items_per_order = df.groupby('order_id').size().reset_index(name='num_items')
```

```python
df1 = df.merge(items_per_order, on='order_id', how='left')
```

```python
df1
```

# Getting frequency

```python
frequency=df1.groupby('customer_unique_id')['order_id'].nunique().reset_index(name='num_of_orders')
```

```python
frequency
```

# Getting Monetary 

```python
monetary=df1.groupby('customer_unique_id')['total_payment_value'].sum().reset_index(name='total_payment')
```

```python
monetary
```

```python
df1.info()
```

# Computing avg delay,avg delivery time,delivery fail rate

```python
df1['order_delivered_customer_date']=pd.to_datetime(df1['order_delivered_customer_date'])
df1['order_purchase_timestamp']=pd.to_datetime(df1['order_purchase_timestamp'])
```

```python
df1['order_estimated_delivery_date']=pd.to_datetime(df1['order_estimated_delivery_date'])
```

```python
df1['delivery_time_days'] = (
    df1['order_delivered_customer_date'] - df1['order_purchase_timestamp']
).dt.days
```

```python
df1['delivery_delay_days'] = (
    df1['order_delivered_customer_date'] - df1['order_estimated_delivery_date']
).dt.days
```

```python
avg_delivery_time = df1.groupby('customer_unique_id')['delivery_time_days'].mean().reset_index(name='avg_delivery_time')
```

```python
avg_delay = df1.groupby('customer_unique_id')['delivery_delay_days'].mean().reset_index(name='avg_delivery_delay')
```

```python
delivery_fail_rate = df.groupby('customer_unique_id')['is_delivered'].mean().reset_index(name='delivery_success_rate')

delivery_fail_rate['delivery_fail_rate'] = 1 - delivery_fail_rate['delivery_success_rate']
```

# Calculating Recency

```python
max_date=df1['order_purchase_timestamp'].max()
```

```python
recency=df1.groupby('customer_unique_id')['order_purchase_timestamp'].max().reset_index(name='last_purchase_date')
```

```python
recency['last_purchase_date']=pd.to_datetime(recency['last_purchase_date'])
```

```python
recency['recency_days']=(max_date-recency['last_purchase_date']).dt.days
```

```python
recency
```

# Computing avg review Score

```python
avg_review_score=df1.groupby('customer_unique_id')['review_score'].mean().reset_index(name='avg_score')
```

```python
avg_review_score
```

```python
avg_items=df1.groupby('customer_unique_id')['num_items'].mean().reset_index(name='avg_items')
```

# Merging all the dataframes into single master customer dataframe

```python
customer_df=(frequency
            .merge(monetary,on='customer_unique_id',how='left')
             .merge(recency[['customer_unique_id','recency_days']],on='customer_unique_id',how='left')
             .merge(avg_review_score,on='customer_unique_id',how='left')
             .merge(avg_items,on='customer_unique_id',how='left')
              .merge(avg_delivery_time, on='customer_unique_id', how='left')
              .merge(avg_delay, on='customer_unique_id', how='left')
             .merge(delivery_fail_rate[['customer_unique_id','delivery_fail_rate']], on='customer_unique_id', how='left')
            )
```

```python
customer_df.head()
```

```python
customer_df.isna().sum()
```

```python
customer_df['avg_delivery_time'] = customer_df['avg_delivery_time'].fillna(customer_df['avg_delivery_time'].median())
customer_df['avg_delivery_delay'] = customer_df['avg_delivery_delay'].fillna(0)
```

```python
customer_df['churned']=(customer_df['num_of_orders']==1).astype(int)
```

```python
customer_df.to_csv('../datasets/customer_df.csv',index=False)
```

```python
customer_df.columns
```

# Modeling the Logistic regression Model

```python
from sklearn.model_selection import train_test_split as tts
```

```python
X=customer_df[['total_payment', 'avg_score', 'avg_items',
       'avg_delivery_time', 'avg_delivery_delay', 'delivery_fail_rate']]
```

```python
y=customer_df['churned']
```

```python
X_train,X_test,y_train,y_test=tts(X,y,train_size=0.85,stratify=y)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
```

```python
from imblearn.over_sampling import SMOTE
sm = SMOTE(random_state=42)

X_train_res, y_train_res = sm.fit_resample(X_train, y_train)
```

```python
logreg=LogisticRegression(class_weight='balanced',max_iter=500,solver='liblinear')
```

```python
logreg.fit(X_train_res,y_train_res)
```

```python
y_proba = logreg.predict_proba(X_test)[:,1]
y_pred = (y_proba > 0.3).astype(int)
```

```python
print(classification_report(y_test, y_pred, target_names=['retained','churned']))
```
